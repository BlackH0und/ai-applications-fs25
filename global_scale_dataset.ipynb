# Global-Scale Dataset (Nov 2024)

The Global-Scale dataset, released **November 23, 2024**, is the newest, largest public resource for satellite-based road extraction. It spans over **13,800 km²**, roughly **20×** more coverage than previous datasets. The dataset provides both segmentation masks and vectorized road graphs for comprehensive evaluation.

## 1. Overview

The dataset collects high-resolution satellite images paired with accurate road graph annotations covering **13,800 km²** worldwide. It is **20×** larger than the previous largest public road extraction benchmark.

## 2. Data & Labels

* **Images:** Multi-source satellite imagery (optical) sampled globally.
* **Annotations:** Both pixel-level segmentation masks and vectorized road graphs, enabling evaluation of segmentation accuracy and topological correctness.
* **Split:** Standard train/validation/test partitions with varied terrains and urban densities.

## 3. Access

The dataset description and annotation protocol are documented in the arXiv preprint *"Towards Satellite Image Road Graph Extraction: A Global-Scale Dataset and A Novel Method"*. The authors plan to release images and vector labels via a public repository.

## 4. Zero-Shot Baseline: SAM-Road (CVPRW 2024)

SAM-Road adapts Meta’s Segment Anything Model to predict dense binary masks for roads and intersections, then extracts graph vertices via non-maximum suppression. It requires no pixel-level fine-tuning beyond the image encoder adaptation and offers a strong zero-shot starting point.

On initial benchmarks, SAM-Road achieves competitive IoU scores relative to fully supervised baselines while requiring minimal annotation effort.

## 5. Comparison with Other Recent Works

* **Sentinel-2 Teacher-Student (Oct 2023):** teacher-student approach generating 50 cm masks from 10 m imagery, achieving 79% mIoU on road detection.
* **Road Damage Dataset (2024):** satellite dataset focusing on potholes and cracks; ~85% F1.
* **MAP-UNet (2023):** upgraded model with multi-head attention; Dice > 0.8 on challenging rural imagery.
